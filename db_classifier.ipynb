```python
import pandas as pd # again alas giving
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
df = pd.read_csv("clean_data.csv") # our clean data
df.dropna(inplace=True) # in upcoming phases there is a error, so we delete nulls agagin
x = df["clean_content"]
y = df["label"]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.25, random_state = 0) # split ourvaribles to 4 x_tarin have our training questions y_train have our training answers x_test is like final and y_test is the answersheet test size says train with 0.75 and rest is for final and last parameter is for the shuffle the datas

print(f"Total rows: {len(df)}")
print(f"Training rows: {len(x_train)}")
print(f"Final rows: {len(x_test)}")
```

```python
 # vectorizing data with tf-idf
vector = TfidfVectorizer()
x_train_tf = vector.fit_transform(x_train)
x_test_tf = vector.transform(x_test) # we are not using fit_ because othervise its use the last learned number

print(f"Shape of x_train_tf: {x_train_tf.shape}")
print(f"Shape of x_test_tf: {x_test_tf.shape}")
```

```python
# initialize and training naive bayes
model_nb = MultinomialNB() # crreat model but empty
model_nb.fit(x_train_tf, y_train) # training part
print("Train done")
```

```python
# waited moment
y_pre_nb = model_nb.predict(x_test_tf) # testing time 
accu_nb = accuracy_score(y_test, y_pre_nb) # finding and initilazing accuracy

print(f"Accuracy: {accu_nb * 100:.2f}%")
print(classification_report(y_test,	y_pre_nb)) # this format is report card
```

```python
# initialize and training logistic regression
model_reg = LogisticRegression(max_iter = 1000) # we limit it with 1000 because if we dont it will keep guessing until it gives an error
model_reg.fit(x_train_tf, y_train) # we will train with exact same data with nb
print("train done")
```

```python
y_pre_reg = model_reg.predict(x_test_tf) # test
accu_reg = accuracy_score(y_test, y_pre_reg) # accuracy

print(f"Accuracy: {accu_reg * 100:.2f}%")
print(classification_report(y_test, y_pre_reg))
```
