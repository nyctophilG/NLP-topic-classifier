```python
import pandas as pd # again alas giving
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
```

```python
df = pd.read_csv("clean_data.csv") # our clean data
df.dropna(inplace=True) # in upcoming phases there is a error, so we delete nulls agagin
x = df["clean_content"]
y = df["label"]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.25, random_state = 42) # split ourvaribles to 4 x_tarin have our training questions y_train have our training answers x_test is like final and y_test is the answersheet test size says train with 0.75 and rest is for final and last parameter is for the shuffle the datas

print(f"Total rows: {len(df)}")
print(f"Training rows: {len(x_train)}")
print(f"Final rows: {len(x_test)}")
```

```python
 # vectorizing data
vector = TfidfVectorizer()
x_train_tf = vector.fit_transform(x_train)
x_test_tf = vector.transform(x_test) # we are not using fit_ because othervise its use the last learned number

print(f"Shape of x_train_tf: {x_train_tf.shape}")
print(f"Shape of x_test_tf: {x_test_tf.shape}")
```

```python
# initialize and training
model_nb = MultinomialNB()
model_nb.fit(x_train_tf, y_train)
print("Train done")
```

```python
# waited moment
y_pre_nb = model_nb.predict(x_test_tf)
accu_nb = accuracy_score(y_test, y_pre_nb)

print(f"Accuracy: {accu_nb * 100:.2f}%")
print(classification_report(y_test,	y_pre_nb))
```
